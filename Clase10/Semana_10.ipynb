# 2. El Laboratorio: "Cocinar los datos" (15 min)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score

# 1. Dataset
data = {
    'canciones_escuchadas': [120, 10, 300, 5, 80, 150, 12, 90],
    'dispositivo': ['Mobile', 'Desktop', 'Mobile', 'Mobile', 'Desktop', 'Desktop', 'Mobile', 'Mobile'],
    'cancelo_suscripcion': [0, 1, 0, 1, 1, 0, 0, 1] # 1 = Se fue, 0 = Se quedó
}
df = pd.DataFrame(data)
print("Datos crudos:\n", df)

#"Miren cómo transformamos 'dispositivo' a números con un solo comando"
df_listo = pd.get_dummies(df, columns=['dispositivo'], drop_first=True)
print("\nDatos transformados (Feature Engineering):\n", df_listo)
```

# 3. El Dilema del Jefe Desconfiado (15 min)

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# Separamos la torta en dos: Entrenamiento y Examen
X = df_listo.drop(columns=['cancelo_suscripcion'])
y = df_listo['cancelo_suscripcion']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# Entrenamos la IA
modelo = LogisticRegression()
modelo.fit(X_train, y_train)

# El examen sorpresa (Test)
predicciones = modelo.predict(X_test)

print(f"¿Pasó el examen? Accuracy en datos nuevos: {accuracy_score(y_test, predicciones) * 100}%")
```

```python
from sklearn.model_selection import cross_val_score

# Usamos el mismo X e y del set de 8 filas anterior
# Evaluamos el modelo usando 4-Fold Cross Validation
scores = cross_val_score(modelo, X, y, cv=4)

print("--- RESULTADOS DE K-FOLD (K=4) ---")
for i, score in enumerate(scores):
    print(f"Examen {i+1}: Accuracy = {score * 100}%")

print("---------------------------------")
print(f"🏆 Rendimiento Promedio Real: {scores.mean() * 100:.2f}%")
```

# 4. El Gran Final: El Despliegue Mínimo Viable (15 min)

```python
import joblib

# Guardamos el cerebro de la IA en un archivo real (.pkl)
joblib.dump(modelo, 'cerebro_streaming_ia.pkl')
print("¡Felicidades! El modelo fue exportado con éxito como 'cerebro_streaming_ia.pkl'.")

# --- SIMULACIÓN DE DESPLIEGUE ---
# Imagina que este código corre en un servidor web meses después:
ia_cargada = joblib.load('cerebro_streaming_ia.pkl')

# Viene un usuario nuevo que escuchó solo 2 canciones y usa Mobile (dispositivo_Mobile = 1)
nuevo_usuario = pd.DataFrame([[2, 1]], columns=['canciones_escuchadas', 'dispositivo_Mobile'])
prediccion_final = ia_cargada.predict(nuevo_usuario)

if prediccion_final[0] == 1:
    print("🚨 Alerta de Retención: ¡Este usuario se va a ir! Ofrecerle descuento ya.")
else:
    print("✅ Usuario fidelizado. Sigue disfrutando del servicio.")
```
