# 📦 Bloque 1: Fundamentos de Matplotlib — Figure, Axes y los Dos Estilos

📘 **Figure, Axes y Axis: la base de todo lo que sigue**

Matplotlib separa tres cosas, y confundirlas es el error más común al empezar:

- **`Figure`**: el contenedor de nivel superior — la imagen completa que se termina mostrando o guardando. Pensala como una diapositiva: hay una sola, pero puede tener uno o varios gráficos puestos arriba.
- **`Axes`**: cada gráfico individual dentro de esa imagen. Es lo que solemos llamar "un gráfico" — tiene su título, su leyenda, sus propios ejes X e Y, sus barras, sus líneas.
- **`Axis`**: la línea graduada (X o Y) que pone las marcas (*ticks*) y los números. Ojo con el nombre: **`Axes`** (el gráfico completo) no es lo mismo que **`Axis`** (un eje puntual). La mayoría de tus cambios se hacen sobre el objeto `ax` (Axes), no sobre `axis`.

Una `Figure` puede tener un solo `Axes` (un gráfico sencillo) o diez (un panel de control complejo, como vamos a ver más adelante con dashboards).

### Los Dos Estilos de Trabajo: Pyplot vs. Orientado a Objetos

Matplotlib ofrece dos formas de escribir el mismo gráfico:

- **Estilo Pyplot (implícito)**: le das órdenes directas a `plt` (`plt.plot()`, `plt.title()`) y Matplotlib asume en qué gráfico querés dibujar. Es rápido para un gráfico único, pero se vuelve confuso apenas hay más de uno en pantalla — es como pintar dos cuadros a la vez usando una sola mano.
- **Estilo Orientado a Objetos — OO (explícito, el recomendado en Data Science)**: creamos variables (`fig`, `ax`) para la Figure y el Axes. Así siempre sabemos exactamente qué objeto estamos modificando, y comparar gráficos lado a lado deja de ser un problema.

El ejemplo de abajo hace el mismo gráfico de las dos formas, para ver la diferencia de código en carne propia.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# ── Estilo Pyplot: le hablamos directo a "plt", sin guardar ninguna referencia ──
plt.plot(x, y, color="steelblue", marker="o")
plt.title("Estilo Pyplot (implícito)")
plt.show()

# ── Estilo Orientado a Objetos: creamos fig y ax, y le hablamos a "ax" ──
fig, ax = plt.subplots()           # fig = la Figure completa; ax = el único Axes adentro
ax.plot(x, y, color="darkorange", marker="o")
ax.set_title("Estilo Orientado a Objetos (explícito)")
plt.show()

# Con UN gráfico, la diferencia no se nota. El problema del estilo Pyplot aparece
# recién cuando hay DOS o más Axes en pantalla: plt.title() le pondría el título
# al último Axes creado, no necesariamente al que vos tenías en mente. Con ax.set_title()
# no hay ambigüedad posible: cada línea le habla a un objeto puntual y conocido.
````

### `subplots()`: Varias Vistas en un Mismo Panel

**¿Para qué sirve?** Para comparar dos cosas relacionadas en una sola imagen: ventas de enero (línea, para ver la evolución semana a semana) al lado de ventas de febrero (barras, para comparar cada semana), los dos dentro del mismo panel. `plt.subplots(nrows, ncols)` crea de una vez la `Figure` y los `Axes` vacíos. A partir de ahí, cada `ax.algo()` dibuja solo en ESE gráfico — así se evita la ambigüedad de no saber dónde está dibujando `plt.plot()` cuando hay más de un gráfico en pantalla. Esto es esencial para el **Análisis Exploratorio de Datos (EDA)**: por ejemplo, en un proyecto inmobiliario, un histograma de precios al lado del scatter metros²-precio.

````python
import matplotlib.pyplot as plt

semanas = ["Sem 1", "Sem 2", "Sem 3", "Sem 4"]

# fig = Figure completa; nrows=1, ncols=2 pide 2 gráficos lado a lado, que se
# desempaquetan directo en ax1 y ax2
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

# ── Gráfico izquierdo (ax1): ventas de Enero, semana a semana ──
ventas_enero = [10, 20, 15, 30]
ax1.plot(
    semanas, ventas_enero,
    color="blue",
    marker="o"      # dibuja un círculo en cada punto de datos, no solo la línea
)
ax1.set_title("Ventas de Enero (miles $)")

# ── Gráfico derecho (ax2): ventas de Febrero, semana a semana ──
ventas_febrero = [12, 18, 22, 25]
ax2.bar(semanas, ventas_febrero, color="orange")  # color = color de relleno de las barras
ax2.set_title("Ventas de Febrero (miles $)")

fig.suptitle("Un solo panel (Figure) con dos gráficos (Axes): Enero vs Febrero", fontsize=14)  # título de TODA la figura
plt.show()
````

### Personalización Completa: de Dibujo a Hallazgo Analítico

Un gráfico sin etiquetas es solo una forma abstracta. Los comandos clave sobre `ax`:

- `ax.set_title()`: el nombre de la historia que estás contando.
- `ax.set_xlabel()` / `ax.set_ylabel()`: qué miden los ejes y en qué unidades — "Precio (USD)", no solo "Precio".
- `ax.legend()`: necesaria en cuanto hay más de una serie en el mismo Axes, para que el lector sepa cuál es cuál.
- `ax.set_xlim()` / `ax.set_ylim()`: vos decidís el rango visible — a veces Matplotlib ajusta los ejes de forma que se pierde perspectiva.
- `fig.tight_layout()`: el salvavidas cuando los títulos de un subplot se encimen con el de al lado.

````python
import matplotlib.pyplot as plt

años = [2020, 2021, 2022, 2023, 2024]
ventas_A = [50, 55, 63, 70, 90]
ventas_B = [48, 52, 58, 65, 68]

fig, ax = plt.subplots(figsize=(8, 5))

# Dibujamos las dos series con etiquetas para la leyenda
ax.plot(años, ventas_A, label='Producto A', color='blue', linewidth=2)
ax.plot(años, ventas_B, label='Producto B', color='orange', linestyle='--')

# Agregamos los elementos de claridad
ax.set_title("Evolución de Ventas Anuales", fontsize=14, fontweight='bold')
ax.set_xlabel("Año")
ax.set_ylabel("Ventas (Miles de Unidades)")
ax.legend()                                       # muestra las etiquetas definidas en 'label'
ax.grid(True, linestyle=':', alpha=0.6)            # una grilla suave ayuda a leer valores
ax.set_ylim(0, 100)                                # fijamos el rango: nada de perspectiva engañosa

plt.tight_layout()
plt.show()
````

⚠️ **Errores comunes a evitar**

- **Confundir `Axes` con `Axis`**: `ax` es el gráfico entero; `axis` es solo la línea de los números.
- **No guardar la referencia a `fig` y `ax`**: si escribís solo `plt.plot()`, perdés el control fino apenas hay más de un gráfico.
- **El "Gráfico de Espagueti"**: meter 10 líneas de colores distintos en un solo `Axes`. Si la visualización es confusa, mejor separar en `subplots`.
- **Olvidar las unidades**: un eje que dice "Ingresos" sin decir si son pesos, dólares o millones es inútil para tomar decisiones.

# 📦 Bloque 2: Análisis Univariado — Histogramas y KDE

📘 **El primer paso de todo análisis: mirar cada variable por separado**

Antes de buscar relaciones entre variables, un data scientist responde 4 preguntas sobre cada variable individual: ¿cuál es el valor más común?, ¿están muy dispersos o concentrados?, ¿hay outliers?, ¿es simétrica o tiene cola hacia un lado?

**Histograma**: divide el rango de la variable en intervalos (*bins*) y cuenta cuántos datos caen en cada uno. La cantidad de bins importa: **pocos bins** dan una imagen demasiado general (pueden ocultar una distribución con dos picos), **demasiados bins** vuelven el gráfico ruidoso, como picos aislados sin patrón.

**KDE** (*Kernel Density Estimation*): en vez de bloques rectangulares, coloca pequeñas "colinas" suaves sobre cada observación y las suma. El resultado es una curva continua que muestra la forma subyacente sin depender de dónde cortamos los bins.

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 4))

# Mismo dato (total_bill), tres cantidades de bins distintas: el mensaje cambia
sns.histplot(data=tips, x="total_bill", bins=5, ax=axes[0])
axes[0].set_title("Muy pocos bins (5): imagen demasiado general")

sns.histplot(data=tips, x="total_bill", bins=60, ax=axes[1])
axes[1].set_title("Demasiados bins (60): ruidoso")

sns.histplot(data=tips, x="total_bill", bins=20, kde=True, ax=axes[2], color="teal")
axes[2].set_title("20 bins + KDE: forma clara")

plt.tight_layout()
plt.show()

# total_bill tiene sesgo a la derecha: la mayoría de las cuentas son bajas/medias,
# pero hay una "cola" de cuentas altas que estira la distribución hacia la derecha.
# Es el mismo patrón que salarios o precios de casas.
````

**Leyendo la forma: sesgo y multimodalidad**

| Patrón | Qué significa |
|---|---|
| **Simetría** | La izquierda es espejo de la derecha (Campana de Gauss). |
| **Sesgo a la derecha** | Cola larga hacia valores altos (ej. salarios, precios de casas — como `total_bill` arriba). |
| **Sesgo a la izquierda** | Cola larga hacia valores bajos. |
| **Multimodalidad** | Dos o más "jorobas" → subgrupos mezclados (ej. alturas de hombres y mujeres juntas en un mismo dataset). |

⚠️ Errores comunes: confundir un histograma con un gráfico de barras (el histograma es para variables numéricas continuas, sin huecos entre barras; el de barras es para categóricas) e ignorar la escala del eje X — un outlier muy lejano puede comprimir todo el resto de los datos en una sola barra.

# 📦 Bloque 3: Seaborn — Distribución y Comparación entre Grupos

📘 **¿Qué es Seaborn y para qué sirve?**

Seaborn **no es "mejor"** que Matplotlib — hace más fáciles los gráficos de análisis de datos. Está construida ENCIMA de Matplotlib, pensada específicamente para graficar DataFrames de Pandas. La diferencia práctica: en Matplotlib puro hay que extraer manualmente cada columna y decirle a `plot()` o `scatter()` qué dibujar, grupo por grupo; en Seaborn se le pasa el DataFrame completo y los nombres de columnas, y arma automáticamente colores, leyenda y estilo.

**Cuándo usar cada uno:**
- **Seaborn** → ideal para análisis exploratorio (EDA): histogramas, boxplots, correlaciones, scatterplots con categorías — la mayoría de los gráficos del día a día.
- **Matplotlib** → cuando necesitás personalizar mucho un gráfico puntual, o hacer una visualización muy específica que Seaborn no tiene resuelta.

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
print(tips.head())

# ── SIN Seaborn: Matplotlib puro necesita separar los datos a mano por categoría ──
fig, ax = plt.subplots(figsize=(6, 4))
for genero, color in [("Male", "steelblue"), ("Female", "orange")]:
    subset = tips[tips["sex"] == genero]                                     # filtrar el DataFrame a mano
    ax.scatter(subset["total_bill"], subset["tip"], color=color, label=genero)  # dibujar cada grupo por separado
ax.legend()                                                                    # la leyenda también se arma a mano
ax.set_title("Con Matplotlib puro: hay que filtrar y dibujar grupo por grupo")
plt.show()

# ── CON Seaborn: una sola línea hace lo mismo (filtra, colorea y arma la leyenda) ──
plt.figure(figsize=(6, 4))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")  # hue= separa por categoría automáticamente
plt.title("Con Seaborn: mismo resultado en 1 línea")
plt.show()
````

**Dos formas en que se comporta Seaborn:** funciones **Axes-level** (`scatterplot`, `boxplot`, `histplot`...) que aceptan `ax=` y dibujan donde nosotros les decimos, y funciones **Figure-level** (`displot`, `relplot`, `catplot`...) que crean su propia imagen y devuelven un `FacetGrid`, no un `Axes`. Se usa Axes-level para integrar el gráfico en un layout propio (varios subplots armados con Matplotlib), y Figure-level para exploración rápida o para dividir automáticamente en paneles por categoría (`col=`).

> ⚠️ Pasar `ax=` a una función figure-level genera error.

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

print("--- 1. EJEMPLO AXES-LEVEL (Obediente) ---")
fig, mi_eje = plt.subplots(figsize=(6, 4))
sns.scatterplot(
    data=tips, x="total_bill", y="tip",
    ax=mi_eje,        # ax= le dice a Seaborn en qué Axes dibujar
    color="purple"
)
mi_eje.set_title("Yo obedezco al 'ax' que creó el programador")
plt.show()

print("\n--- 2. EJEMPLO FIGURE-LEVEL (Independiente) ---")
g = sns.displot(
    data=tips,
    x="total_bill",
    col="time",   # separa el gráfico en un panel por cada valor de "time": 2 gráficos automáticos
    kind="kde",
    fill=True
)
plt.show()
````

### Boxplot y Violinplot: Comparando Grupos

Un **Boxplot** resume una distribución en 5 números clave: **Mínimo**, **Q1 (25%)**, **Mediana (Q2)**, **Q3 (75%)** y **Máximo** — la caja es el rango intercuartílico (IQR, el 50% central), y los puntos fuera de los bigotes son outliers (¡no siempre un error de carga: puede ser tu cliente más valioso!).

Un **Violinplot** combina el boxplot con una curva KDE: el ancho representa la frecuencia. Dos grupos pueden tener la misma mediana y los mismos cuartiles (boxplots iguales) pero formas de distribución totalmente distintas — el violín las revela.

- Usá **Boxplot** si tu audiencia no es técnica o solo te interesa variabilidad y extremos.
- Usá **Violinplot** si necesitás la forma precisa y tenés suficientes registros.

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(11, 4.5))

sns.boxplot(data=tips, x="time", y="total_bill", ax=axes[0])       # 5 números + outliers
axes[0].set_title("Boxplot: resumen por categoría")

sns.violinplot(data=tips, x="time", y="total_bill", ax=axes[1])    # boxplot + forma completa
axes[1].set_title("Violinplot: boxplot + forma")

plt.tight_layout()
plt.show()
````

### Heatmap de Correlación

Convierte una tabla de números en una cuadrícula de colores. El uso más común: visualizar la **matriz de correlación** (coeficiente de Pearson, entre -1 y 1).

- **Cerca de 1** (color intenso): las variables se mueven juntas (ej. metros² y precio).
- **Cerca de -1** (color intenso opuesto): se mueven en direcciones contrarias (ej. peso del auto y eficiencia).
- **Cerca de 0** (color neutro): no hay relación clara.

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

# Solo tiene sentido correlacionar columnas numéricas
matriz_corr = tips[["total_bill", "tip", "size"]].corr()

plt.figure(figsize=(5, 4))
sns.heatmap(matriz_corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)  # annot=True: muestra el número en cada celda
plt.title("Heatmap de correlación: total_bill, tip, size")
plt.show()

# total_bill y tip correlacionan fuerte y positivo (cuentas más caras -> propinas más altas):
# es exactamente el tipo de relación que un heatmap deja ver de un vistazo, sin calcular
# cada par de variables a mano.
````

**Estrategia: ¿qué gráfico elegir?**

| Pregunta | Gráfico |
|---|---|
| ¿Outliers o dispersión en un grupo? | Boxplot |
| ¿Forma y densidad de la distribución? | Violinplot |
| ¿Qué variables están relacionadas? | Heatmap |

⚠️ Errores comunes: saturar el heatmap con 50 variables (filtrá primero las más importantes), violinplots sobre grupos de 5 datos (engañoso — mejor un boxplot o un stripplot), y no ordenar las categorías de un boxplot de mayor a menor mediana (reduce la carga cognitiva de quien lo lee).

# 📦 Bloque 4: Storytelling y Diseño de Dashboards

📘 **En breve**

Antes de escribir código, dos preguntas evitan la mayoría de los gráficos malos: **¿el diseño ayuda a leer el dato o lo estorba?** y **¿el resultado es honesto y accesible, o manipula lo que se ve?** Los dos ejemplos que siguen bajan estas ideas a código, cada uno con su propia explicación.

**Sobre las librerías:** el Ejemplo 1 usa Matplotlib (imagen estática, gráfico de barras). El Ejemplo 2 combina Plotly (gráfico interactivo en HTML) con Matplotlib de nuevo — no hace falta saber cómo funciona cada una por dentro para seguir el ejemplo, alcanza con mirar el resultado.

### Ejemplo 1 — Encuadre, Jerarquía Visual y Anotaciones

Mismos datos, dos gráficos. A la izquierda, el resultado de graficar sin pensarlo (`ax.bar()` a secas). A la derecha, el mismo `ax.bar()` con tres decisiones de diseño encima: título con la unidad (encuadre), un color distinto para el mes que importa (jerarquía visual) y una flecha que explica por qué (anotación).

````python
import matplotlib.pyplot as plt

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ventas = [40, 38, 45, 42, 50, 48, 47, 53, 55, 60, 62, 80]

fig, (ax_sin_diseno, ax_con_diseno) = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

# ── SIN diseño: un solo color, título sin unidad, no queda claro qué mirar ──
ax_sin_diseno.bar(meses, ventas, color="steelblue")
ax_sin_diseno.set_title("ventas")  # título genérico: no dice la unidad ni qué destacar

# ── CON diseño: encuadre (título + unidad) + jerarquía visual (color) + anotación ──
colores = ['#AAAAAA'] * 11 + ['#1a5276']  # gris para los primeros 11 meses, azul oscuro para diciembre
ax_con_diseno.bar(meses, ventas, color=colores)

ax_con_diseno.set_title("Ventas mensuales 2024 (miles de $)")
ax_con_diseno.set_ylabel("Ventas (miles $)")

ax_con_diseno.annotate(
    'Récord del año',
    xy=(11, 80),
    xytext=(7, 68),
    arrowprops=dict(arrowstyle='->')
)

plt.tight_layout()
plt.show()
````

### Ejemplo 2 — Ética Visual: el Mismo Dato, Dos Historias Distintas

La primera versión (torta con Plotly) separa el sector de Apple y titula "35% Apple" — visualmente parece que domina el mercado. La segunda versión (barras horizontales) ordena por valor real: "Otras Compañías" (40%) queda primera, más que Apple. Los números son idénticos en ambas — lo que cambia es el diseño.

````python
import plotly.express as px
import pandas as pd

data = {
    'Compañía': ['Apple', 'Google', 'Microsoft', 'Otras Compañías'],
    'Cuota de Mercado (%)': [35, 15, 10, 40]
}
df = pd.DataFrame(data)

fig = px.pie(
    df,
    values='Cuota de Mercado (%)',
    names='Compañía',
    title='Cuota de Mercado Global de Tecnología (35% Apple)',  # título que ya sesga el mensaje hacia Apple
    hole=0.4,
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig.update_traces(
    textinfo='percent+label',
    pull=[0.15, 0, 0, 0]  # separa el sector de Apple del resto para que parezca más dominante
)
fig.show()
````

````python
import matplotlib.pyplot as plt

# Mismos datos que el gráfico anterior, ordenados de mayor a menor cuota
empresas = ['Otras Compañías', 'Apple', 'Google', 'Microsoft']
cuotas = [40, 35, 15, 10]

fig, ax = plt.subplots(figsize=(7, 3))
ax.barh(empresas, cuotas, color='#5DADE2')  # barras horizontales, todas del mismo color
ax.set_xlabel('Cuota de mercado (%)')
ax.set_title('Cuota de mercado — sin destacar ninguna empresa')
ax.set_xlim(0, 50)  # el eje arranca en 0: regla de honestidad para gráficos de barras

for i, v in enumerate(cuotas):
    ax.text(v + 0.5, i, f'{v}%', va='center')  # muestra el valor exacto al lado de cada barra

plt.tight_layout()
plt.show()
````

**¿Cuál de los dos es mejor? El de barras.** No porque el pie chart esté mal hecho técnicamente — el problema es que `pull` + el título elegido comunican una conclusión que los datos no sostienen: "Apple domina", cuando en realidad "Otras Compañías" tiene más cuota (40% vs 35%). Un gráfico honesto no es el más lindo, es el que un lector puede interpretar bien sin ayuda de quien lo hizo. Nada de esto reemplaza el chequeo de accesibilidad cromática (paleta `colorblind`, que vemos en el Bloque 6) — ética y accesibilidad van de la mano.

### El Eje Y Truncado: la Trampa Más Común

Empezar el eje vertical en un número distinto a cero para exagerar una diferencia. Puede hacer que un crecimiento del 1% parezca un crecimiento del 50%. Mantené siempre el cero como base, salvo que haya una razón estadística muy fuerte para no hacerlo — y si la hay, aclaralo explícitamente en el gráfico.

````python
import matplotlib.pyplot as plt

trimestres = ["Q1", "Q2", "Q3", "Q4"]
ingresos = [1000, 1005, 1008, 1012]  # una variación mínima, apenas ~1.2% en todo el año

fig, (ax_truncado, ax_honesto) = plt.subplots(nrows=1, ncols=2, figsize=(11, 4))

# ── Eje truncado: arranca en 995, casi al ras de las barras ──
ax_truncado.bar(trimestres, ingresos, color="#E74C3C")
ax_truncado.set_ylim(995, 1015)  # ¡acá está la trampa!
ax_truncado.set_title("Eje truncado: parece un salto enorme")
ax_truncado.set_ylabel("Ingresos (miles $)")

# ── Eje honesto: arranca en 0, la variación real se ve minúscula ──
ax_honesto.bar(trimestres, ingresos, color="#27AE60")
ax_honesto.set_ylim(0, 1200)
ax_honesto.set_title("Eje desde 0: la variación real (~1.2%)")
ax_honesto.set_ylabel("Ingresos (miles $)")

plt.tight_layout()
plt.show()

# Mismos 4 números en los dos gráficos. El de la izquierda "grita" una tendencia
# dramática que, en la escala real (derecha), es casi imperceptible.
````

### Los Pilares del Storytelling y del Dashboard

**Cuatro preguntas, cuatro tipos de gráfico:**

| Pilar | Pregunta | Ejemplo de gráfico |
|---|---|---|
| **Distribución** | ¿Cómo se reparten mis datos? | Histograma, KDE |
| **Relación** | ¿Depende X de Y? | Scatter, Heatmap |
| **Evolución temporal** | ¿Crecemos, decrecemos o estancamos? | Línea |
| **Composición** | ¿Qué porcentaje viene de cada parte? | Barras apiladas |

**Narrativa (Data Storytelling):** como todo relato, un análisis visual tiene **Inicio** (contexto: "las ventas fluctuaron"), **Nudo** (el problema: "cae fuerte en la región norte") y **Desenlace** (la acción: "investigar al nuevo competidor").

**El "semáforo cognitivo" del color:**
- **Secuenciales**: variables numéricas (azul claro→oscuro para ingresos bajos→altos).
- **Divergentes**: punto medio crítico (rojo/blanco/azul para temperatura bajo/en/sobre cero).
- **Categóricos**: grupos sin orden — pero respetá las convenciones (rojo = pérdida, no al revés; si no, confundís al 100% de tu audiencia).

**Regla de oro del dashboard:** máximo 5 visualizaciones principales — más que eso genera ruido visual y confunde en vez de aclarar.

**Casos reales:** *Retail* — heatmaps de recorrido de clientes (+15% ventas moviendo productos de baja rotación a zonas de alto tráfico visual). *Salud pública* — áreas apiladas para ver capacidad hospitalaria antes del punto de quiebre. *Amazon* — una caída súbita en "añadir al carrito" delata un bug técnico en segundos, no una pérdida de interés real. *Finanzas personales* — donuts de gastos, fáciles de leer aunque el pie chart sea criticable en contextos técnicos.

# 📦 Bloque 5: Arquitectura Visual y la Regla de Lectura en "Z"

📘 **El dashboard como interfaz de usuario**

Un dashboard en Excel (o en cualquier herramienta de BI) es una interfaz de usuario, no una colección de gráficos insertados en una hoja de cálculo. El objetivo no es solo mostrar datos, sino **reducir la carga cognitiva**: el usuario debe poder extraer los insights más críticos en los primeros 3 segundos.

En las culturas occidentales leemos de izquierda a derecha y de arriba hacia abajo. En pantallas sin mucho texto continuo, el ojo humano tiende a escanear la información siguiendo una ruta en forma de la letra **"Z"**. Al construir un dashboard, conviene distribuir los elementos clave exactamente sobre ese recorrido imaginario.

| Posición | Qué colocar ahí |
|---|---|
| **1. Arriba-Izquierda** (Ancla visual) | KPIs principales: Ventas Totales, Margen, ROI. |
| **2. Arriba-Derecha** (Contexto y control) | Slicers horizontales, línea de tiempo (Timeline). |
| **3. Abajo-Izquierda** (Profundidad analítica) | Tendencias en el tiempo, comparativas clave. |
| **4. Abajo-Derecha** (Punto de salida) | Tablas de detalle, top 10, formato condicional. |

Recorrido recomendado: **KPIs → Controles/Slicers → Gráficos de Tendencia → Tablas de Detalle**. Ocultar las líneas de cuadrícula y usar el tamaño de las celdas como sistema de grillas ayuda a que el ojo siga ese recorrido sin distracciones.

*(Este módulo es conceptual — se aplica al diseñar dashboards en Excel/BI, no tiene una celda de código Python asociada.)*

# 📦 Bloque 6: Principios de UI/UX y Jerarquía Visual en Dashboards

📘 **¿Qué es la jerarquía visual?**

Imaginá la cabina de un avión: cientos de indicadores compitiendo por tu atención. Un piloto entrenado sabe dónde mirar; un ejecutivo con 5 minutos antes de una junta, no. La **jerarquía visual** es el orden en que el ojo humano percibe lo que ve — el sistema que le dice al usuario "mirá esto primero, esto después, esto solo si necesitás el detalle". Sin jerarquía, todos los elementos "pesan" igual: el resultado es ruido visual. El cerebro es perezoso por naturaleza y busca atajos — un diseño con jerarquía clara permite que el insight salte a la vista sin esfuerzo.

**Los pilares del diseño visual en reporting:**

| Pilar | Idea clave |
|---|---|
| **Patrón F/Z** | Zona superior izq. = KPIs críticos. Zona media = tendencias. Zona inferior/derecha = detalle. |
| **Tamaño y peso** | Lo más grande se percibe como más importante. Fuentes Sans Serif (Segoe UI, Aptos) para números. |
| **Color con propósito** | Semánticos: rojo = alerta, verde = cumplimiento, ámbar = atención. Grises para ejes y etiquetas. |
| **Espacio en blanco** | No es espacio desperdiciado: agrupa conceptos relacionados y separa secciones. |

💻 **Ejemplo de Código Sencillo**

El mismo principio de "menos es más" se aplica a un gráfico de Matplotlib: cada borde, sombra o línea de más es *ruido* que compite con el dato.

````python
import matplotlib.pyplot as plt

categorias = ["Norte", "Sur", "Este", "Oeste", "Centro"]
valores = [82, 65, 90, 58, 74]
colores_random = ["#E74C3C", "#3498DB", "#F1C40F", "#9B59B6", "#1ABC9C"]  # un color distinto por barra, sin motivo

fig, (ax_chartjunk, ax_limpio) = plt.subplots(nrows=1, ncols=2, figsize=(12, 4.5))

# ── "Chartjunk": bordes gruesos, colores random sin significado, grilla pesada ──
ax_chartjunk.bar(categorias, valores, color=colores_random, edgecolor="black", linewidth=2.5)
ax_chartjunk.set_title("Chartjunk: mucho ruido, poco dato", fontsize=13)
ax_chartjunk.grid(True, linewidth=1.5, color="black")
for spine in ax_chartjunk.spines.values():
    spine.set_linewidth(2.5)

# ── Limpio: un solo color, sin bordes de más, grilla apenas visible ──
ax_limpio.bar(categorias, valores, color="#4C72B0")
ax_limpio.set_title("Data-Ink Ratio alto: solo lo necesario", fontsize=13)
ax_limpio.spines["top"].set_visible(False)     # sacamos los bordes que no aportan información
ax_limpio.spines["right"].set_visible(False)
ax_limpio.grid(axis="y", alpha=0.3)             # grilla suave, solo horizontal, de fondo

plt.tight_layout()
plt.show()

# Mismos 5 valores en los dos gráficos. Edward Tufte llama a esto maximizar el
# "Data-to-Ink Ratio": si podés quitar un elemento y el dato se sigue entendiendo
# igual de bien, quitalo.
````

**Arquitectura de la información: de lo general a lo particular**

Un dashboard ejecutivo es como una conversación: **El Titular (KPIs)** — "¿cómo vamos?"; **El Contexto (Tendencias)** — "¿cómo llegamos hasta acá?"; **El Diagnóstico (Desgloses)** — "¿quién o qué causó esto?"; **El Detalle (Tablas)** — "mostrame los datos exactos".

*Ejemplo real — dashboard de un Controller:* Nivel 1: tarjetas de EBITDA, Ingresos, Costos y % Desviación. Nivel 2: gráfico de cascada (Waterfall). Nivel 3: top 5 centros de costo excedidos. Nivel 4: tabla dinámica filtrable con facturas específicas.

**Errores comunes y "anti-patrones":**

| Anti-patrón | Solución |
|---|---|
| **Efecto Árbol de Navidad** | Demasiados colores vibrantes compitiendo. Usá paleta monocromática y reservá el color brillante para excepciones. |
| **Chartjunk** (Tufte) | Bordes gruesos, sombras 3D, líneas oscuras — visto en el ejemplo de arriba. |
| **Dato Huérfano** | "$1.000.000" sin contexto no dice nada. Acompañá cada KPI con un comparativo (vs. Target, vs. Año Anterior) o un Sparkline. |

### Accesibilidad: No la Dejes para el Final

~8% de los hombres tiene algún tipo de daltonismo (el más común: confunde rojo y verde). Depender solo del semáforo rojo/verde es un riesgo real. Solución: respaldar siempre el color con `style=` (forma o trazo) y usar `palette="colorblind"`.

````python
import matplotlib.pyplot as plt
import seaborn as sns

flights = sns.load_dataset("flights")

df_meses = flights[flights['month'].isin(['Jan', 'Jul', 'Dec'])]        # nos quedamos con 3 meses para no saturar el gráfico
df_meses['month'] = df_meses['month'].cat.remove_unused_categories()    # limpia categorías vacías tras el filtro

plt.figure(figsize=(8, 4))
sns.lineplot(
    data=df_meses,
    x="year",
    y="passengers",
    hue="month",           # color por mes
    style="month",          # trazo distinto por mes (respaldo del color)
    markers=True,           # marcador geométrico distinto por mes
    palette="colorblind"    # paleta apta para daltonismo
)
plt.title("Accesible: Diferentes colores, trazos y figuras")
plt.show()
````

## A. GridSpec: cuando `subplots()` no alcanza

📘 **Pensalo así**

Un **subplot** es un gráfico dentro de una figura. `subplots()` crea varios gráficos juntos en una grilla. Por ejemplo, `plt.subplots(2, 2)` arma una ventana con 4 gráficos, todos del mismo tamaño:

```
+-------+-------+
|   1   |   2   |
+-------+-------+
|   3   |   4   |
+-------+-------+
```

**¿Qué pasa si uno tiene que ser más importante?** Ahí aparece `GridSpec`. En vez de pedir una grilla pareja, le decís cómo repartir el espacio. Por ejemplo, `height_ratios=[1, 3]` significa "la fila de arriba mide 1 parte, la de abajo mide 3 partes":

```
+-----------------------+
|   gráfico pequeño     |
+-----------------------+
|                       |
|    gráfico grande     |
|                       |
+-----------------------+
```

**En resumen:**
- `subplots()` → cuando todos los gráficos pueden tener el mismo tamaño.
- `GridSpec` → cuando algunos gráficos tienen que ser más grandes o más chicos que otros, como en un dashboard.

El ejemplo de código que sigue arma justamente un mini-dashboard: arriba (chico) un resumen rápido de la tendencia trimestral, abajo (grande, ratio 3) el gráfico principal con el detalle por región.

````python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(8, 6))  # figsize=(8, 6): tamaño del lienzo completo en pulgadas

# 2 filas, 1 columna. height_ratios=[1, 3]: la fila de abajo mide 3 veces la de arriba
gs = gridspec.GridSpec(nrows=2, ncols=1, height_ratios=[1, 3], hspace=0.4)

ax_resumen = fig.add_subplot(gs[0])    # slot 0: fila de arriba (chica) — el "resumen" del dashboard
ax_principal = fig.add_subplot(gs[1])  # slot 1: fila de abajo (grande) — el gráfico principal

# ── Arriba (chico): resumen rápido, tendencia trimestral ──
trimestres = ["Q1", "Q2", "Q3", "Q4"]
ventas_trimestre = [120, 145, 130, 180]
ax_resumen.plot(trimestres, ventas_trimestre, color="red", marker="o", linewidth=2)
ax_resumen.set_title("Resumen: ventas por trimestre (miles $)")

# ── Abajo (grande): el gráfico principal, detalle por región ──
regiones = ["Norte", "Sur", "Este", "Oeste"]
ventas_region = [95, 60, 150, 80]
ax_principal.bar(regiones, ventas_region, color="steelblue")
ax_principal.set_title("Detalle: ventas por región — gráfico principal del dashboard")
ax_principal.set_ylabel("Ventas (miles $)")

plt.show()
````

## B. Series Temporales: Fechas, Resampling y Rolling Window

📘 **Recordatorio: ¿qué es una serie temporal?**

Un conjunto de datos donde cada valor está asociado a un momento en el tiempo, ordenado cronológicamente. Lo que la distingue de una tabla común es que **el orden importa**. Por eso Pandas necesita saber que esa columna es realmente una fecha (`datetime`) y no texto.

📘 **En breve**

Pandas lee las fechas de un CSV como texto (`object`) salvo que le digamos lo contrario. Si graficamos texto en el eje X, **no se reordena solo**. `pd.to_datetime(col, format=...)` convierte el texto a un tipo `datetime64[ns]` que sí sabe ordenarse solo y graficarse correctamente.

> ⚠️ Si hay fechas mal formateadas, agregá `errors='coerce'` para que se conviertan a `NaT` en vez de tirar error.

💻 Ejemplo de Código Sencillo — imaginemos que tenemos un registro donde los años se cargaron desordenados o como texto:

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Años desordenados y cargados como texto (strings) — fijate el orden: 2025, 2023, 2026, 2024
data = {'Año': ['2025', '2023', '2026', '2024'], 'Ventas': [50, 30, 60, 40]}
df = pd.DataFrame(data)

print("--- 1. GRÁFICO INCORRECTO (Fechas como texto) ---")
plt.figure(figsize=(5, 3))
sns.lineplot(data=df, x="Año", y="Ventas", marker="o")  # x="Año" sigue siendo texto (dtype object)
plt.title("Arruinado: sigue el orden de la tabla (2025, 2023, 2026, 2024)")
plt.show()

print("\n--- 2. GRÁFICO CORRECTO (Convertido a Datetime) ---")
df['Año_Correcto'] = pd.to_datetime(df['Año'], format='%Y')  # convierte el texto a datetime64[ns]
plt.figure(figsize=(10, 3))
sns.lineplot(data=df, x="Año_Correcto", y="Ventas", marker="o")  # ahora ordena cronológicamente solo
plt.title("Cronología Perfecta usando Datetime")
plt.show()
````

````python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import seaborn as sns

# (df ya tiene la columna 'Año_Correcto' como datetime de la celda anterior)
plt.figure(figsize=(10, 3))
sns.lineplot(data=df, x="Año_Correcto", y="Ventas", marker="o")

ax = plt.gca()  # gca() = "get current axes", el eje activo del gráfico actual
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))  # cómo se muestra cada etiqueta (solo el año)
ax.xaxis.set_major_locator(mdates.YearLocator())            # dónde va cada marca (una por año)

plt.title("Cronología Perfecta mostrando solo el Año")
plt.show()
````

### Resampling y Rolling Window: Suavizando la Serie

📘 **Pensalo con el precio del dólar todos los días**

Sube y baja todo el tiempo, cuesta ver la tendencia de fondo. Dos formas de "alisarla":

- **`resample('W').mean()`**: agrupa los días por semana y calcula el promedio de cada una — desaparece gran parte del ruido diario.
- **`rolling(window=3).mean()`**: no agrupa, va promediando las últimas 3 observaciones, y la "ventana" avanza de a una.

Es muy común usar los dos juntos: primero `resample()` (agrupa y reduce el ruido diario), después `rolling()` sobre esa serie ya semanal (suaviza todavía más la tendencia).

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)  # fija la semilla para que el resultado sea reproducible

# Precio simulado del dólar: 90 días, con ruido diario alrededor de $1000
fechas = pd.date_range("2024-01-01", periods=90, freq="D")
precio_dolar = np.random.normal(loc=1000, scale=50, size=90).round()
precio_dolar[45] += 150  # evento puntual: una corrida cambiaria dispara el precio ese día

serie = pd.Series(precio_dolar, index=fechas)  # el índice es la fecha, no un número de fila

semanal = serie.resample("W").mean()          # agrupa los 90 días en ~13 semanas y promedia
rolling_3 = semanal.rolling(window=3).mean()  # media móvil de 3 semanas sobre la serie ya semanal

plt.figure(figsize=(10, 4))
plt.plot(serie.index, serie, alpha=0.3, label="Precio diario (con ruido)")
plt.plot(semanal.index, semanal, marker="o", label="Promedio semanal (resample)")
plt.plot(rolling_3.index, rolling_3, linewidth=2, label="Rolling 3 semanas")

fecha_evento = fechas[45]
plt.annotate(
    "Corrida cambiaria",
    xy=(fecha_evento, serie[fecha_evento]),
    xytext=(fecha_evento, serie[fecha_evento] + 100),
    arrowprops=dict(arrowstyle="->")
)

plt.legend()
plt.title("Precio del dólar: diario vs. suavizado semanal y rolling")
plt.show()
````

## C. Análisis Multivariado: Canales Estéticos

📘 **En breve**

Un gráfico 2D solo tiene X e Y. Para mostrar más variables, las mapeamos a **canales estéticos**: `hue` (color, categórico), `style` (forma/trazo, categórico), `size` (tamaño, numérico). La posición sigue siendo el canal más preciso; la forma, el más débil. Con 4-5 variables cruzadas el gráfico se vuelve difícil de leer — mejor dividir en paneles (`col=`) o pasar a Plotly.

💻 Ejemplo de Código Sencillo — usando el dataset de propinas (tips), cruzamos 4 variables en un solo gráfico plano: cuenta total (Eje X), propina (Eje Y), ¿fumador o no? (Color/Hue) y tamaño de la mesa (Tamaño/Size).

````python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=tips,
    x="total_bill",   # posición X
    y="tip",          # posición Y
    hue="smoker",     # color por categoría
    size="size",      # tamaño según valor numérico
    sizes=(20, 200)   # rango de tamaños en píxeles: mesa chica ≈ 20, mesa grande ≈ 200
)
plt.title("Análisis Multivariado: 4 variables en 2 dimensiones")
plt.xlabel("Total de la Cuenta ($)")
plt.ylabel("Propina ($)")
plt.show()
````

## E. Formatos de Exportación: Rasterizado vs Vectorial

📘 **¿Qué significa "rasterizado"?**

Una imagen rasterizada se guarda como una grilla de píxeles. El archivo no sabe qué es "un círculo" o "una línea" — solo sabe "en la posición (120, 45) hay un píxel rojo". Por eso, al ampliar mucho, se "pixela". `.png` y `.jpg` son formatos rasterizados.

Lo opuesto es **vectorial**: guarda instrucciones matemáticas, recalculables para cualquier tamaño sin perder nitidez. `.pdf` y `.svg` son formatos vectoriales.

**¿Y el DPI?** *Dots per inch*: cuántos píxeles se imprimen por pulgada de papel. `dpi=300` es el estándar profesional para impresión; `dpi=72` es lo típico para pantalla.

**Regla práctica:** PNG a 300dpi para presentaciones/web, PDF para documentos impresos.

💻 Ejemplo de Código Sencillo — guardamos el mismo gráfico en ambos formatos:

````python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import seaborn as sns

penguins = sns.load_dataset("penguins")

sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g")
plt.title("Gráfico para Exportar")

# savefig() guarda A UN ARCHIVO todo lo que ya está dibujado en la figura activa.
# Tiene que ir SIEMPRE antes de plt.show() o plt.close(), que "vacían" la figura activa.
plt.savefig("grafico_alta_resolucion.png", dpi=300)   # rasterizado, 300dpi = estándar profesional
plt.savefig("grafico_vectorial_infinito.pdf")          # mismo gráfico, ahora vectorial

plt.close()
print("¡Archivos guardados en tu panel izquierdo de Colab/Jupyter!")

plt.figure(figsize=(6, 4))
plt.imshow(mpimg.imread("grafico_alta_resolucion.png"))
plt.axis("off")
plt.title("Así quedó el PNG guardado (300dpi)")
plt.show()
````

### El parámetro `bbox_inches='tight'`

📘 **En breve**

`bbox_inches='tight'` recalcula el encuadre de la imagen guardada para que títulos largos o etiquetas rotadas no queden cortados. Distinto de `plt.tight_layout()`, que ajusta el espaciado entre subgráficos en pantalla — conviene usar ambos juntos.

💻 Ejemplo de Código Sencillo — forzamos un título largo para ver la diferencia entre guardar con y sin este parámetro:

````python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import seaborn as sns

penguins = sns.load_dataset("penguins")

sns.histplot(data=penguins, x="flipper_length_mm", color="teal")
plt.title("Este es un título ridículamente largo que se suele cortar al exportar", fontsize=14, pad=20)
plt.xlabel("Largo de la Aleta (mm)", fontsize=14, labelpad=20)

plt.savefig("grafico_mal_encuadrado.png")                       # sin tight: puede cortar el título/eje
plt.savefig("grafico_perfecto.png", bbox_inches='tight')        # con tight: encuadre perfecto

plt.close()

fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

ax1.imshow(mpimg.imread("grafico_mal_encuadrado.png"))
ax1.set_title("SIN bbox_inches='tight' (título cortado)")
ax1.axis("off")

ax2.imshow(mpimg.imread("grafico_perfecto.png"))
ax2.set_title("CON bbox_inches='tight' (encuadre perfecto)")
ax2.axis("off")

plt.tight_layout()
plt.show()
````
